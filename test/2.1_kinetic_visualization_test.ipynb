# Inhibition Model Visualization Demo

This notebook demonstrates how to visualize inhibition factor profiles for kinetic models in AnimaKinetica.

## Features:
- Plot all inhibition profiles in a single multi-panel figure
- Plot individual inhibition factors
- Visualize nutrient factor (Monod) profiles
- Analyze combined effects during simulation

---

```python
# Import necessary modules
import numpy as np
from animakinetica.core.parameters import SimulationParameters
from animakinetica.core.simulation import CellCultureSimulation
from animakinetica.models.kinetics import create_kinetic_model, Kinetics
from animakinetica.models.stoichiometry import StoichiometricModel
from animakinetica.models.odesystems import OperatingMode
from animakinetica.analysis.kinetic_visualization import (
    plot_inhibition_profiles,
    plot_single_inhibition,
    plot_nutrient_factors,
    plot_kinetics,
    plot_simulated_IFs,
    plot_simulated_NFs,
    plot_kinetic_rates
)
```

## 1. Initialize Kinetic Model

First, load parameters and create a kinetic model instance.

```python
# Load default parameters
params = SimulationParameters.load_parameters(["base_parameters","cell_GFI","anab_GFI", "cata_GFI"])
params.update("excess_kinetics_batch")  

# Create stoichiometric model
stoich_model = StoichiometricModel(params)

# Create kinetic model (EXCESS kinetics has more parameters)
kinetic_model = create_kinetic_model(Kinetics.EXCESS, params, stoich_model)

print(f"Kinetic Model Type: {type(kinetic_model).__name__}")
print(f"\nInhibition Models:")
print(f"  Lactate: {kinetic_model.inhibition_model_Lac}")
print(f"  Ammonia: {kinetic_model.inhibition_model_Amm}")
print(f"  Osmolality: {kinetic_model.inhibition_model_Osm}")
print(f"\nKey Parameters:")
print(f"  IC100_Lac: {kinetic_model.IC100_Lac:.2f} mmol/L")
print(f"  IC100_Amm: {kinetic_model.IC100_Amm:.2f} mmol/L")
print(f"  IC50_O2: {kinetic_model.IC50_O2:.4f} mmol/L")
print(f"  IC100_CO2: {kinetic_model.IC100_CO2:.2f} mmol/L")
```

## 2. Plot All Inhibition Profiles

Visualize all inhibition factors in a comprehensive multi-panel figure.

```python
# Plot all inhibition profiles
fig = plot_inhibition_profiles(
    kinetic_model,
    n_points=200,
    show_ic_markers=True
)

fig.show()
```

## 3. Plot Individual Inhibition Factors

For detailed analysis, plot individual inhibition profiles.

```python
# Lactate inhibition
fig_lac = plot_single_inhibition(
    kinetic_model,
    inhibitor='lactate',
    concentration_range=(0, 120),
    show_ic_markers=True
)
fig_lac.show()
```

```python
# Ammonia inhibition
fig_amm = plot_single_inhibition(
    kinetic_model,
    inhibitor='ammonia',
    concentration_range=(0, 30),
    show_ic_markers=True
)
fig_amm.show()
```

```python
# Osmolality inhibition (shows mirroring around baseline)
fig_osm = plot_single_inhibition(
    kinetic_model,
    inhibitor='osmolality',
    show_ic_markers=True
)
fig_osm.show()
```

## 4. Plot Nutrient Factors (Monod Kinetics)

Visualize how substrate availability affects growth through Monod kinetics.

```python
# Plot nutrient factors
fig_nf = plot_nutrient_factors(
    kinetic_model,
    n_points=200,
    show_k_markers=True
)

fig_nf.show()
```

## 5. Analyze Combined Effects During Simulation

Run a simulation and visualize how nutrient and inhibition factors change over time.

```python
# Load default parameters
params = SimulationParameters.load_parameters(["base_parameters","cell_GFI","anab_GFI", "cata_GFI"])
params.update("excess_kinetics_batch")
# Create and run simulation with default settings
sim = CellCultureSimulation(
    parameters=params,
)

results = sim.run_simulation()
```

```python
# This creates a comprehensive overview of all nutrient factors, inhibition factors, and kinetic rates
fig_kinetics = plot_kinetics(sim, batch_number=1)
fig_kinetics.show()
# Or using another handle
# fig_kinetics_method2 = sim.data.plot_kinetics(batch_number=1)
# fig_kinetics_method2.show()
```

### 5.1. Plot Nutrient Factors Over Time

Visualize how nutrient factors (glucose, glutamine, O2) evolve during the batch, showing substrate limitation effects.

```python
# Plot nutrient factors time course
fig_nf_sim = plot_simulated_NFs(sim, batch_number=1)
fig_nf_sim.show()

# Alternative: using the data object method
# fig_nf_sim = sim.data.plot_nutrient_factors(batch_number=1)
# fig_nf_sim.show()
```

### 5.2. Plot Inhibition Factors Over Time

Visualize how inhibition factors (lactate, ammonia, O2, CO2, osmolality) change during the batch, showing when growth inhibition becomes significant.

```python
# Plot inhibition factors time course
fig_if_sim = plot_simulated_IFs(sim, batch_number=1)
fig_if_sim.show()

# Alternative: using the data object method
# fig_if_sim = sim.data.plot_inhibition_factors(batch_number=1)
# fig_if_sim.show()
```

### 5.3. Plot Kinetic Rates Over Time

Visualize how specific growth rate (μ) and substrate uptake rates (q-rates) evolve during the batch, showing the dynamic response to changing conditions.

```python
# Plot kinetic rates time course
fig_rates = plot_kinetic_rates(sim, batch_number=1)
fig_rates.show()

# Alternative: using the data object method
# fig_rates = sim.data.plot_kinetic_rates(batch_number=1)
# fig_rates.show()
```

### 5.4. Understanding the Complete Kinetics Picture

The three plots above show different aspects of the same simulation:

- **Nutrient Factors (NF)**: How substrate availability limits growth (Monod kinetics)
  - Values approach 0 when substrates are depleted
  - Values approach 1 when substrates are abundant
  
- **Inhibition Factors (IF)**: How metabolic waste products inhibit growth
  - Values approach 0 when inhibition is severe
  - Values approach 1 when there's no inhibition
  
- **Kinetic Rates**: The resulting specific rates from the combined effects
  - Growth rate μ = μ_max × NF × IF
  - Substrate uptake rates (q-rates) follow similar patterns

These visualizations help identify:
- **Bottlenecks**: Which nutrient or inhibitor limits growth at different stages
- **Optimization targets**: Which parameters to adjust for better performance
- **Process dynamics**: How the cell culture responds to changing conditions

## 6. Compare Different Inhibition Models

Compare how different inhibition model types affect the profile shape.

```python
import plotly.graph_objects as go

# Create models with different inhibition types
inhibition_models = ['linear', 'exponential', 'hyperbolic', 'power_decay']
colors = ['#E74C3C', '#3498DB', '#2ECC71', '#F39C12']

fig = go.Figure()

# Use lactate parameters as example
IC50 = kinetic_model.IC50_Lac
IC100 = kinetic_model.IC100_Lac
phi = kinetic_model.phi_power_decay_Lac

C_range = np.linspace(0, IC100 * 1.2, 200)

for model_type, color in zip(inhibition_models, colors):
    # Calculate IF for each model type
    IF_values = []
    for c in C_range:
        try:
            if_val = kinetic_model.inhibition_factor(
                c, model_type,
                IC50=IC50,
                IC100=IC100,
                phi_power_decay=phi
            )
            IF_values.append(max(0, if_val))  # Clamp to 0
        except:
            IF_values.append(0)
    
    fig.add_trace(
        go.Scatter(
            x=C_range, y=IF_values,
            mode='lines',
            name=model_type.capitalize(),
            line=dict(color=color, width=2)
        )
    )

fig.update_layout(
    title="Comparison of Inhibition Model Types",
    xaxis_title="Inhibitor Concentration (mmol/L)",
    yaxis_title="Inhibition Factor",
    yaxis_range=[0, 1.1],
    height=500,
    width=800,
    template='plotly_white',
    hovermode='x unified',
    legend=dict(x=0.7, y=0.95)
)

fig.show()
```

## 7. Sensitivity to IC Parameters

Visualize how changing IC50/IC100 values affects the inhibition profile.

```python
import plotly.graph_objects as go

# Test different IC100 values for lactate
IC100_values = [50, 75, 100, 125]  # Different IC100 values
colors = ['#E74C3C', '#E67E22', '#F39C12', '#F1C40F']

fig = go.Figure()

for IC100_test, color in zip(IC100_values, colors):
    C_range = np.linspace(0, IC100_test * 1.2, 200)
    # Use centralized inhibition_factor method instead of hardcoding
    IF_values = [
        max(0, kinetic_model.inhibition_factor(
            c, 
            kinetic_model.inhibition_model_Lac,
            IC50=kinetic_model.IC50_Lac, 
            IC100=IC100_test,
            phi_power_decay=kinetic_model.phi_power_decay_Lac
        ))
        for c in C_range
    ]
    
    fig.add_trace(
        go.Scatter(
            x=C_range, y=IF_values,
            mode='lines',
            name=f'IC100 = {IC100_test} mmol/L',
            line=dict(color=color, width=2)
        )
    )

fig.update_layout(
    title=f"Effect of IC100 on Lactate Inhibition (φ = {kinetic_model.phi_power_decay_Lac})",
    xaxis_title="Lactate Concentration (mmol/L)",
    yaxis_title="Inhibition Factor",
    yaxis_range=[0, 1.1],
    height=500,
    width=800,
    template='plotly_white',
    hovermode='x unified',
    legend=dict(x=0.65, y=0.95)
)

fig.show()
```
